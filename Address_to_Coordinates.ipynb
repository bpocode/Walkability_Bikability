# Address to coordinates

Run the installation cell once, then edit `address` in the next cell and run it.

This simple test uses OpenStreetMap's Nominatim service through geopy. No API key is needed; an internet connection is required. Your address is sent to the service. Check the matched address to confirm the result.

For occasional single-address tests only; wait at least one second between lookups.

[geopy documentation](https://geopy.readthedocs.io/en/stable/) · [© OpenStreetMap contributors](https://www.openstreetmap.org/copyright)

```python
import ssl
import certifi
from geopy.geocoders import Nominatim
from geopy.exc import GeocoderServiceError

# Change this address, then run this cell.
address = "1 Grand Avenue, San Luis Obispo, CA, USA"

# Use a trusted certificate bundle for the secure connection.
ssl_context = ssl.create_default_context(cafile=certifi.where())
geolocator = Nominatim(
    user_agent="bikability-address-test/1.0",
    timeout=15,
    ssl_context=ssl_context,
)
latitude = longitude = None

try:
    location = geolocator.geocode(address)
    if location is None:
        print("No match found. Try adding the city, state, and postal code.")
    else:
        latitude, longitude = location.latitude, location.longitude
        print("Matched address:", location.address)
        print("Latitude:", latitude)
        print("Longitude:", longitude)
except GeocoderServiceError as error:
    print(f"Lookup failed. Check your internet connection or try again later. ({error})")
```
