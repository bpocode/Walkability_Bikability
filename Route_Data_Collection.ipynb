# Data Collection

Run from this notebook's folder. The first loop gets routes and 100-foot walking or 150-foot biking points (plus turns) with elevations.
The second loop downloads photos and saves a coordinate-to-image CSV in each route folder.
Both walking and biking CSVs in `Test Routes` are included.

Your key is loaded from `.env`. Running collection makes Google API calls; saved routes and photos are reused.
Missing photos are recorded as unavailable in JSON; no nearby fallback image is returned. All saving and mapping details live in `master.py`.


```python
from master import read_routes, get_route_data, save_route_images, plot_routes
from IPython.display import display

PITCH = -5  # Camera tilt in degrees, used for both walking and biking.
```

```python
# Get each route from the CSVs and add it to a list.
route_list = []

for route in read_routes():
    print(f"Getting {route['travel_mode']} route {route['route_id']}", flush=True)
    route_data = get_route_data(route)
    route_list.append(route_data)
```

```python
# Get the photos for each route and save them in its own folder.
for route_data in route_list:
    print(f"Getting photos for {route_data['travel_mode']} {route_data['route_id']}", flush=True)
    save_route_images(route_data, pitch=PITCH)
```
